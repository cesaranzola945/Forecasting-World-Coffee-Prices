```python
# load SPEI
import pandas as pd
dfw = pd.read_csv(path_dat + "spei_agg.csv")
```

```python
# load arabica coffee price
dfp = pd.read_csv(path_dat + "arabica_coffee.csv")
```

```python
# laod coffee region spei 
dfw = pd.read_csv(path_dat + "spei_agg.csv")
```

```python
# merge both variables
df = pd.merge(dfw,dfp,how='inner',on=('year','month'))
```

```python
df['date_t'] = pd.to_datetime(df['date_t'])
```

```python
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

fig,ax1 = plt.subplots(figsize=(10,7))
l1, = ax1.plot(df['date_t'],df['COFFEE_ARABIC'], label='Coffee Price',color='green')
ax1.set_ylabel('USD per Lb')

ax2 = ax1.twinx()
l2, ax2.plot(df['date_t'],df['spei12'], label='SPEI',color='blue')
ax2.set_ylabel('SPEI Index')

plt.setp(ax1.get_xticklabels(), rotation=90)
fig.legend(handles=[l1, l2], loc="upper left")

plt.show()
```
